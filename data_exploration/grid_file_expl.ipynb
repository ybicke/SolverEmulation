```python
import xarray as xr
import numpy as np

def analyze_grid_connectivity(ds, cell_index=0):
    """Analyze connectivity for a specific cell"""
    # Get neighbors
    neighbors = ds['neighbor_cell_index'].values[:, cell_index]
    cell_loc = (ds['clon'].values[cell_index], ds['clat'].values[cell_index])
    
    print(f"Cell {cell_index+1} analysis:")
    print(f"Location: lon={cell_loc[0]:.4f}, lat={cell_loc[1]:.4f} (radians)")
    print(f"Neighbors (1-based indexing): {neighbors}")
    
    # Get parent cell
    parent = ds['parent_cell_index'].values[cell_index]
    print(f"Parent cell: {parent}")
    
    # Find edges connected to this cell
    edges = ds['edge_of_cell'].values[:, cell_index]
    print(f"Connected edges: {edges}")
    
    # Get distances to neighbors through these edges
    for i, edge in enumerate(edges):
        distances = ds['edge_cell_distance'].values[:, edge-1]  # -1 for 0-based indexing
        print(f"Edge {edge} distances: {distances[0]:.2f}m, {distances[1]:.2f}m")

# Usage:
ds = xr.open_dataset("/mydata/deepcloud/yves/SolverEmulation/data_exploration/icon_grid_0008_R02B05_G.nc")
analyze_grid_connectivity(ds, cell_index=0)
```

```python
import xarray as xr
import numpy as np

def analyze_grid_structure(ds):
    print("=== GRID STRUCTURE ANALYSIS ===\n")
    
    # 1. Cell Location Analysis
    print("1. CELL LOCATIONS (clon, clat)")
    for var in ['clon', 'clat']:
        print(f"\n{var.upper()}:")
        print(f"Dimensions: {ds[var].dims}")
        print(f"Shape: {ds[var].shape}")
        print(f"Units: {ds[var].attrs.get('units', 'not specified')}")
        print(f"Range: [{ds[var].min().values:.4f}, {ds[var].max().values:.4f}] radians")
        print("Attributes:", ds[var].attrs)
        print(f"First 5 values: {ds[var].values[:5]}")
    
    # 2. Neighbor Cell Index Analysis
    print("\n2. NEIGHBOR RELATIONSHIPS (neighbor_cell_index)")
    neighbor = ds['neighbor_cell_index']
    print(f"Dimensions: {neighbor.dims}")
    print(f"Shape: {neighbor.shape}")
    print(f"Attributes: {neighbor.attrs}")
    print("\nExample neighbor structure:")
    print(f"First cell neighbors: {neighbor.values[:, 0]}")  # All neighbors of first cell
    print(f"Second cell neighbors: {neighbor.values[:, 1]}")  # All neighbors of second cell
    
    # 3. Parent-Child Structure Analysis
    print("\n3. PARENT-CHILD RELATIONSHIPS")
    parent_child_vars = [var for var in ds.variables if 'parent' in var or 'child' in var]
    print(f"\nFound {len(parent_child_vars)} parent/child related variables: {parent_child_vars}")
    
    for var in parent_child_vars:
        print(f"\n{var.upper()}:")
        print(f"Dimensions: {ds[var].dims}")
        print(f"Shape: {ds[var].shape}")
        print(f"Attributes: {ds[var].attrs}")
        print(f"First 5 values: {ds[var].values[:5]}")
        
        # Special analysis for parent_cell_index
        if var == 'parent_cell_index':
            unique_parents = np.unique(ds[var].values)
            print(f"Number of unique parent cells: {len(unique_parents)}")
            print(f"Parent cell IDs range: [{min(unique_parents)}, {max(unique_parents)}]")
            
            # Count children per parent
            values, counts = np.unique(ds[var].values, return_counts=True)
            print("\nChildren per parent distribution:")
            for val, count in zip(values[:5], counts[:5]):  # Show first 5 parents
                print(f"Parent {val}: {count} children")
    
    # 4. Edge Cell Distance Analysis
    print("\n4. EDGE CELL DISTANCE ANALYSIS")
    edge_dist = ds['edge_cell_distance']
    print(f"Dimensions: {edge_dist.dims}")
    print(f"Shape: {edge_dist.shape}")
    print(f"Units: {edge_dist.attrs.get('units', 'not specified')}")
    print(f"Attributes: {edge_dist.attrs}")
    print("\nDistance statistics:")
    print(f"Min distance: {float(edge_dist.min()):.2f} meters")
    print(f"Max distance: {float(edge_dist.max()):.2f} meters")
    print(f"Mean distance: {float(edge_dist.mean()):.2f} meters")
    print(f"First few values:")
    print(edge_dist.values[:, :5])  # First 5 edges
    
    # 5. Grid Level Information
    print("\n5. GRID LEVEL INFORMATION")
    print("Grid Attributes:")
    grid_attrs = ['grid_level', 'grid_root', 'grid_ID', 'parent_grid_ID', 'number_of_grid_used']
    for attr in grid_attrs:
        if attr in ds.attrs:
            print(f"{attr}: {ds.attrs[attr]}")
    
    # 6. Refinement Information
    print("\n6. REFINEMENT VARIABLES")
    refin_vars = [var for var in ds.variables if 'refin' in var]
    print(f"Found {len(refin_vars)} refinement variables: {refin_vars}")
    for var in refin_vars:
        print(f"\n{var}:")
        print(f"Dimensions: {ds[var].dims}")
        print(f"Shape: {ds[var].shape}")
        print(f"Attributes: {ds[var].attrs}")

def main():
    # Load the dataset
    file_path = "/mydata/deepcloud/yves/SolverEmulation/data_exploration/icon_grid_0008_R02B05_G.nc"
    ds = xr.open_dataset(file_path)
    
    try:
        analyze_grid_structure(ds)
    finally:
        ds.close()

if __name__ == "__main__":
    main()
```

```python
import xarray as xr
import numpy as np
import pandas as pd
from collections import defaultdict

def analyze_parent_child_hierarchy(ds):
    print("=== DETAILED PARENT-CHILD HIERARCHY ANALYSIS ===\n")
    
    parent_cell_index = ds['parent_cell_index'].values
    
    # 1. Create parent-child mapping
    parent_to_children = defaultdict(list)
    for child_idx, parent_idx in enumerate(parent_cell_index, start=1):  # 1-based indexing
        parent_to_children[parent_idx].append(child_idx)
    
    # 2. Basic Statistics
    print("BASIC HIERARCHY STATISTICS:")
    print(f"Total number of cells: {len(parent_cell_index)}")
    print(f"Number of parent cells: {len(parent_to_children)}")
    print(f"Children per parent: {len(parent_cell_index) / len(parent_to_children):.0f}")
    
    # 3. Detailed Parent Analysis
    print("\nPARENT DISTRIBUTION ANALYSIS:")
    children_counts = [len(children) for children in parent_to_children.values()]
    print(f"Min children per parent: {min(children_counts)}")
    print(f"Max children per parent: {max(children_counts)}")
    print(f"Average children per parent: {np.mean(children_counts):.2f}")
    
    # 4. Show first few parent-child relationships in detail
    print("\nDETAILED PARENT-CHILD RELATIONSHIPS (First 5 parents):")
    for parent_id in sorted(list(parent_to_children.keys()))[:5]:
        children = parent_to_children[parent_id]
        print(f"\nParent {parent_id}:")
        print(f"  Number of children: {len(children)}")
        print(f"  Child indices: {children}")
        
        # Show coordinates for parent's children
        child_lons = ds['clon'].values[np.array(children) - 1]  # -1 for 0-based indexing
        child_lats = ds['clat'].values[np.array(children) - 1]
        print("  Children coordinates (lon, lat in radians):")
        for i, (lon, lat) in enumerate(zip(child_lons, child_lats)):
            print(f"    Child {children[i]}: ({lon:.4f}, {lat:.4f})")
    
    # 5. Analyze the grid hierarchy levels
    print("\nGRID HIERARCHY INFORMATION:")
    print(f"Grid level: {ds.attrs.get('grid_level', 'Not specified')}")
    print(f"Grid root: {ds.attrs.get('grid_root', 'Not specified')}")
    print(f"Number of grids used: {ds.attrs.get('number_of_grid_used', 'Not specified')}")
    
    # 6. Analyze refinement structure
    if 'index_c_list' in ds:
        print("\nREFINEMENT LEVEL STRUCTURE:")
        index_c_list = ds['index_c_list'].values
        print("Start indices for refinement levels:", index_c_list[0])
        print("End indices for refinement levels:", index_c_list[1])
    
    # 7. Neighbor relationships within parent cells
    print("\nNEIGHBOR RELATIONSHIPS WITHIN PARENT CELLS:")
    neighbor_cell_index = ds['neighbor_cell_index'].values
    
    # Analyze first parent's children and their neighbors
    first_parent_children = parent_to_children[1]
    print(f"\nAnalyzing children of first parent (Parent 1):")
    for child in first_parent_children:
        child_neighbors = neighbor_cell_index[:, child-1]  # -1 for 0-based indexing
        print(f"  Child {child} neighbors: {child_neighbors}")
        # Check which neighbors belong to same parent
        same_parent_neighbors = [n for n in child_neighbors if n in first_parent_children]
        print(f"    Neighbors in same parent cell: {same_parent_neighbors}")

def main():
    file_path = "/mydata/deepcloud/yves/SolverEmulation/data_exploration/icon_grid_0008_R02B05_G.nc"
    ds = xr.open_dataset(file_path)
    
    try:
        analyze_parent_child_hierarchy(ds)
    finally:
        ds.close()

if __name__ == "__main__":
    main()
```

```python
import xarray as xr
import numpy as np

def analyze_refinement_levels_simple(ds):
    """
    Provide a high-level overview of the refinement hierarchy.
    """
    parent_cell_index = ds['parent_cell_index'].values
    total_cells = len(parent_cell_index)
    
    print(f"Total number of cells: {total_cells}")
    print(f"Grid level: {ds.attrs.get('grid_level', 'Not specified')}")
    
    # Find unique parent cells
    unique_parents = np.unique(parent_cell_index)
    
    # Find root cells (cells that are their own parents)
    root_cells = unique_parents[unique_parents == np.arange(1, len(unique_parents) + 1)]
    print(f"\nNumber of root cells: {len(root_cells)}")
    
    # Estimate number of refinement levels
    estimated_levels = int(np.log(total_cells / len(root_cells)) / np.log(4)) + 1
    print(f"\nEstimated number of refinement levels: {estimated_levels}")
    
    # Print refinement ratio
    print(f"\nRefinement ratio: 1:4 (each cell is divided into 4 cells at the next level)")

def main():
    file_path = "/mydata/deepcloud/yves/SolverEmulation/data_exploration/icon_grid_0008_R02B05_G.nc"
    ds = xr.open_dataset(file_path)
    
    try:
        analyze_refinement_levels_simple(ds)
    finally:
        ds.close()

if __name__ == "__main__":
    main()
```
