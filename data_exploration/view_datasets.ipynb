```python
import xarray as xr

def show_radiation_dimensions(file_path):
    """Show dimensions of radiation tendency variables"""
    print(f"\n=== Radiation Tendencies Dimensions ===")
    print(f"File: {file_path}")
    
    # Open the NetCDF file
    ds = xr.open_dataset(file_path)
    
    variables = ['ddt_temp_radlw', 'ddt_temp_radsw']
    
    for var_name in variables:
        if var_name in ds:
            var = ds[var_name]
            print(f"\n{var_name}:")
            print(f"  Dimensions: {var.dims}")
            print(f"  Shape: {var.shape}")
            print(f"  Height dimension size: {var.sizes['height_2']}")
            print(f"  Height values: {ds['height_2'].values}")
    
    ds.close()

# Analyze the file
file_path = '../../../../../mydata/deepcloud/salman/dataset/tmp/ml_ecrad_ape_R2B05_myrunscript_1year_183min_atm_3d_DOM01_ml_0001_lonlat.nc'
show_radiation_dimensions(file_path)
```

```python
import xarray as xr
import numpy as np

def investigate_height_dimension(file_path):
    """Investigate the height dimension of flux variables"""
    ds = xr.open_dataset(file_path)
    
    print("=== Height Dimension Investigation ===")
    
    # Check height coordinate values directly
    print("\nHeight coordinate values:")
    print(f"height values: {ds['height'].values}")
    print(f"height size: {len(ds['height'].values)}")
    
    # Check actual data dimensions for flux variables
    flux_vars = ['lwflx_up', 'lwflx_dn', 'swflx_up', 'swflx_dn']
    
    for var_name in flux_vars:
        print(f"\n{var_name}:")
        var = ds[var_name]
        print(f"  Variable shape: {var.shape}")
        print(f"  Actual number of height levels in data: {var.shape[1]}")
        
        # Check if there are any missing values or NaN at the boundaries
        print(f"  First height level has data: {not np.all(np.isnan(var.isel(time=0, height=0)))}")
        print(f"  Last height level has data: {not np.all(np.isnan(var.isel(time=0, height=-1)))}")
    
    ds.close()

# Run the investigation
file_path = '../../../../../mydata/deepcloud/salman/dataset/tmp/ml_ecrad_ape_R2B05_myrunscript_1year_183min_atm_3d_DOM01_ml_0001_lonlat.nc'
investigate_height_dimension(file_path)
```

```python
import xarray as xr

def show_dataset_info(file_path):
    """Show dataset information including shapes and selected values"""
    ds = xr.open_dataset(file_path)
    
    # Print global attributes
    print("Global Attributes:")
    for attr in ds.attrs:
        print(f"{attr}: {ds.attrs[attr]}")

    # Print dimensions
    print("\nDimensions:")
    for dim, size in ds.dims.items():
        print(f"{dim}: size {size}")

    # Print variables and selected attributes, including shapes and values
    print("\nVariables and Selected Descriptions:")
    for var in ds.data_vars:
        print(f"\n{var}:")
        print(f"  Dimensions: {ds[var].dims}")
        print(f"  Shape: {ds[var].shape}")
        print(f"  Size: {ds[var].size}")
        if 'standard_name' in ds[var].attrs:
            print(f"    Name: {ds[var].attrs['long_name']}")
        if 'units' in ds[var].attrs:
            print(f"    Units: {ds[var].attrs['units']}")
        if 'param' in ds[var].attrs:
            print(f"    Param: {ds[var].attrs['param']}")
        
        # Print selected values for albvisdir and albnirdir
        if var in ['albvisdir', 'albnirdir']:
            print("  Selected Values:")
            print(f"    First time step, first 5 cells: {ds[var].values[0, :10]}")
            print(f"    Last time step, last 5 cells: {ds[var].values[-1, -10:]}")

    ds.close()

# Analyze the file
file_path = '../../../../../mydata/deepcloud/salman/dataset/tmp/ml_ecrad_ape_R2B05_myrunscript_1year_183min_atm_3d_DOM01_ml_0001_lonlat.nc'
show_dataset_info(file_path)
```

```python
# Define the time index for which you want to plot the data
time_index = 100  # Adjust this index to select a different time point
num_levels = 50  # Adjust the number of contour levels

# List of variables to plot
variables = [
    'temp',           # Temperature
    'qi',             # Specific cloud ice content
    'qc',             # Specific cloud water content
    'qv',             # Specific humidity
    'clc',            # Cloud cover
    'pres',           # Pressure
]

num_plots = len(variables)
num_cols = 2
num_rows = (num_plots + 1) // num_cols

fig, axes = plt.subplots(
    num_rows, num_cols, 
    figsize=(24, 8 * num_rows), 
    subplot_kw={'projection': ccrs.PlateCarree()}
)

for i in range(num_plots):
    var = variables[i]
    row = i // num_cols
    col = i % num_cols
    # Handle case when axes is 1D or 2D array
    if num_rows == 1:
        ax = axes[col]
    else:
        ax = axes[row, col]
    
    # Check the dimensions of the variable
    if 'height' in ds[var].dims:
        # If the variable depends on 'height', average over 'height'
        avg_data = ds[var].isel(time=time_index).mean(dim='height')
    elif 'height_2' in ds[var].dims:
        avg_data = ds[var].isel(time=time_index).mean(dim='height_2')
    else:
        # If the variable does not depend on 'height', use as is
        avg_data = ds[var].isel(time=time_index)
    
    avg_data_np = avg_data.values.flatten()
    
    # Handle missing values
    if np.isnan(avg_data_np).all():
        print(f"All data is NaN for variable '{var}' at time index {time_index}. Skipping plot.")
        continue

    # Handle variables with constant values
    data_min = np.nanmin(avg_data_np)
    data_max = np.nanmax(avg_data_np)
    if not np.isfinite(data_min) or not np.isfinite(data_max) or data_min == data_max:
        print(f"Variable '{var}' at time index {time_index} has constant or invalid values. Skipping plot.")
        continue
    
    levels = np.linspace(data_min, data_max, num=num_levels)
    
    # Get longitude and latitude (convert to degrees if necessary)
    clon = ds['clon'].values
    clat = ds['clat'].values
    # If 'clon' and 'clat' are in radians, convert to degrees
    # clon = np.degrees(ds['clon'].values)
    # clat = np.degrees(ds['clat'].values)
    
    # Create the contour plot
    contourf = ax.tricontourf(
        clon, clat, avg_data_np, 
        levels=levels, transform=ccrs.PlateCarree(), cmap='coolwarm'
    )
    
    # Add a colorbar
    cbar = fig.colorbar(
        contourf, ax=ax, orientation='vertical', fraction=0.025, pad=0.1
    )
    cbar.set_label(
        f"Average {ds[var].attrs.get('long_name', var)} ({ds[var].attrs.get('units', '')})"
    )
    
    # Set plot title
    ax.set_title(
        f"Average accross heights {ds[var].attrs.get('long_name', var)} at Time Index {time_index}"
    )
    
    # Add gridlines
    gl = ax.gridlines(draw_labels=True, dms=True, x_inline=False, y_inline=False)
    gl.top_labels = False
    gl.right_labels = False

# Remove any unused subplots
total_subplots = num_rows * num_cols
if total_subplots > num_plots:
    for i in range(num_plots, total_subplots):
        row = i // num_cols
        col = i % num_cols
        if num_rows == 1:
            fig.delaxes(axes[col])
        else:
            fig.delaxes(axes[row, col])

plt.tight_layout()
plt.show()
plt.close()
```
